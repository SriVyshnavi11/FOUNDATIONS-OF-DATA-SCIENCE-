```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("Titanic-Dataset (1).csv")

# 1. MISSING VALUES
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# 2. REDUNDANT VALUES
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

# 3. OUTLIERS
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df["Fare"] < Q1 - 1.5*IQR) |
              (df["Fare"] > Q3 + 1.5*IQR)]

print("Outliers:")
print(outliers)

# 4. STANDARDIZATION
scaler = StandardScaler()
df["Fare_Standardized"] = scaler.fit_transform(df[["Fare"]])

print(df[["Fare", "Fare_Standardized"]].head())

# 5. GROUPING
grouped = df.groupby("Sex")["Age"].mean()

print("Average Age by Gender:")
print(grouped)

# 6. RESHAPING
reshaped = df.pivot_table(
    values="Fare",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
)

print("Reshaped Data:")
print(reshaped)

# 7. FILTERING
filtered = df[(df["Age"] > 18) & (df["Fare"] > 50)]

print("Filtered Data:")
print(filtered.head())
```
