```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Titanic-Dataset.csv')

print(df.head())
print(df.shape)
print(df.columns)
```

```python
# BASIC DATA INSPECTION

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 10 rows:")
display(df.head(10))

print("\nLast 10 rows:")
display(df.tail(10))

print("\nRandom 5 rows:")
display(df.sample(5))

print("\nUnique values:")
for col in df.columns:
    print(col, ":", df[col].nunique())
```

```python
# MISSING VALUE HANDLING

print("Missing values before handling:")
print(df.isnull().sum())

# Fill Age with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill Embarked with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Fill Cabin with Unknown
df['Cabin'] = df['Cabin'].fillna('Unknown')

print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python
# DUPLICATE HANDLING

print("Number of duplicate rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)

print("\nDuplicate rows after removal:")
print(df.duplicated().sum())
```

```python
# FILTERING USING LOC AND ISIN

# Passengers whose Age is greater than 30
print("Passengers with Age > 30:")
display(df.loc[df['Age'] > 30, ['Name', 'Age', 'Sex', 'Pclass']])

# Passengers from Pclass 1 or 2
print("\nPassengers from Pclass 1 or 2:")
display(df.loc[df['Pclass'].isin([1, 2]), ['Name', 'Pclass', 'Fare']])

# Female passengers
print("\nFemale passengers:")
display(df.loc[df['Sex'].isin(['female']), ['Name', 'Sex', 'Age']])
```

```python
# RENAMING COLUMNS

df.rename(columns={
    'PassengerId': 'Passenger_ID',
    'Pclass': 'Passenger_Class',
    'SibSp': 'Siblings_Spouses',
    'Parch': 'Parents_Children'
}, inplace=True)

print("Updated column names:")
print(df.columns)

display(df.head())
```

```python
# DATA TYPE CONVERSION

print("Data types before conversion:")
print(df.dtypes)

# Convert selected columns
df['Passenger_ID'] = df['Passenger_ID'].astype(int)
df['Passenger_Class'] = df['Passenger_Class'].astype(int)
df['Survived'] = df['Survived'].astype(int)

print("\nData types after conversion:")
print(df.dtypes)
```

```python
# HANDLING EXTRA SPACES IN STRING DATA

# Remove leading/trailing spaces from string columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

print("Spaces removed from string columns.")

# Check sample values
display(df[['Name', 'Sex', 'Embarked']].head())
```

```python
# GROUPBY OPERATIONS

print("1. Survival Rate by Sex:")
print(df.groupby('Sex')['Survived'].mean())

print("\n2. Survival Rate by Passenger Class:")
print(df.groupby('Passenger_Class')['Survived'].mean())

print("\n3. Number of Passengers by Passenger Class:")
print(df.groupby('Passenger_Class').size())

print("\n4. Average Age by Passenger Class:")
print(df.groupby('Passenger_Class')['Age'].mean())

print("\n5. Average Fare by Passenger Class:")
print(df.groupby('Passenger_Class')['Fare'].mean())

print("\n6. Survival Rate by Sex and Passenger Class:")
print(df.groupby(['Sex', 'Passenger_Class'])['Survived'].mean())

print("\n7. Multiple Statistics:")
display(df.groupby('Passenger_Class').agg({
    'Age': ['mean', 'min', 'max'],
    'Fare': ['mean', 'min', 'max'],
    'Survived': 'mean'
}))

print("\n8. Statistics by Sex:")
display(df.groupby('Sex').agg({
    'Age': 'mean',
    'Fare': 'mean',
    'Survived': 'mean'
}))
```

```python
# OUTLIER DETECTION USING IQR

Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[(df['Fare'] < lower_limit) | (df['Fare'] > upper_limit)]

print("\nNumber of Fare outliers:", len(outliers))

display(outliers[['Name', 'Passenger_Class', 'Fare']])
```

```python
# BOXPLOT

plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Fare'])
plt.title('Fare Outlier Detection')
plt.xlabel('Fare')
plt.show()
```

```python
# Z-SCORE OUTLIER DETECTION

from scipy.stats import zscore

df['Fare_ZScore'] = zscore(df['Fare'])

zscore_outliers = df[
    (df['Fare_ZScore'] > 3) |
    (df['Fare_ZScore'] < -3)
]

print("Number of Z-score outliers:", len(zscore_outliers))

display(zscore_outliers[['Name', 'Fare', 'Fare_ZScore']])
```

```python
# MIN-MAX NORMALIZATION

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df['Fare_Normalized'] = scaler.fit_transform(df[['Fare']])

print("Original and Normalized Fare:")
display(df[['Fare', 'Fare_Normalized']].head(10))
```

```python
# STANDARDIZATION

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df['Fare_Standardized'] = scaler.fit_transform(df[['Fare']])

print("Original Fare and Standardized Fare:")
display(df[['Fare', 'Fare_Standardized']].head(10))
```

```python
# CATEGORICAL DATA ENCODING

# Label encoding for Sex
df['Sex_Encoded'] = df['Sex'].map({
    'male': 0,
    'female': 1
})

# One-hot encoding for Embarked
df = pd.get_dummies(
    df,
    columns=['Embarked'],
    prefix='Embarked',
    dtype=int
)

print("Dataset after encoding:")
display(df.head())

print("\nUpdated columns:")
print(df.columns.tolist())
```

```python
# CORRELATION ANALYSIS

numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

print("Correlation Matrix:")
display(correlation)

plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# FINAL DATA WRANGLING CHECK

print("Final Dataset Shape:", df.shape)

print("\nFinal Columns:")
print(df.columns.tolist())

print("\nFinal Data Types:")
print(df.dtypes)

print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nFinal Dataset:")
display(df.head(10))

print("\nStatistical Summary:")
display(df.describe())
```

```python
# SAVE FINAL WRANGLED DATASET

output_file = '/content/Titanic_Wrangled.csv'

df.to_csv(output_file, index=False)

print("Wrangled dataset saved successfully!")
print("File:", output_file)
```
