```python
import pandas as pd
```

```python
import pandas as pd

employee = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
               111, 112, 113, 114, 115],

    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma',
             'Frank', 'Grace', 'Henry', 'Isha', 'Jack',
             'Kiran', 'Liam', 'Maya', 'Noah', 'Olivia'],

    'Department': ['HR', 'IT', 'Sales', 'Finance', 'IT',
                   'HR', 'Marketing', 'IT', 'Finance', 'Sales',
                   'HR', 'IT', 'Marketing', 'Finance', 'Sales']
})

print(employee)
```

```python
salary = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
               116, 117, 118, 119, 120],

    'Designation': ['Manager', 'Developer', 'Sales Executive',
                    'HR', 'Developer', 'HR Executive',
                    'Marketing Manager', 'Software Engineer',
                    'Financial Analyst', 'Sales Manager',
                    'Data Analyst', 'Team Lead', 'Consultant',
                    'Business Analyst', 'Project Manager'],

    'Salary': [50000, 60000, 45000, 55000, 65000,
               48000, 70000, 75000, 58000, 62000,
               68000, 80000, 72000, 67000, 85000]
})

print(salary)
```

```python
print("EMPLOYEE DATAFRAME")
print(employee)

print("\nSALARY DATAFRAME")
print(salary)
```

```python
merged = pd.merge(employee, salary, on='Emp_ID')

print(merged)
```

```python
# 1. Inner Merge
print("\nINNER MERGE")
inner = pd.merge(employee, salary, on='Emp_ID', how='inner')
print(inner)


# 2. Outer Merge
print("\nOUTER MERGE")
outer = pd.merge(employee, salary, on='Emp_ID', how='outer')
print(outer)


# 3. Left Merge
print("\nLEFT MERGE")
left = pd.merge(employee, salary, on='Emp_ID', how='left')
print(left)


# 4. Right Merge
print("\nRIGHT MERGE")
right = pd.merge(employee, salary, on='Emp_ID', how='right')
print(right)
```

```python
# doin concatination vertically and horizontlly
# Employee DataFrame 2 - 15 records
employee2 = pd.DataFrame({
    'Emp_ID': [116, 117, 118, 119, 120, 121, 122, 123, 124, 125,
               126, 127, 128, 129, 130],

    'Name': ['Aarav', 'Bhavya', 'Chetan', 'Divya', 'Eshan',
             'Farah', 'Gaurav', 'Harini', 'Ishaan', 'Jhanvi',
             'Karthik', 'Lavanya', 'Manoj', 'Nandini', 'Pranav'],

    'Department': ['IT', 'HR', 'Sales', 'Finance', 'Marketing',
                   'IT', 'HR', 'Finance', 'Sales', 'Marketing',
                   'IT', 'Finance', 'HR', 'Sales', 'IT']
})

print("Employee DataFrame 2:")
print(employee2)

```

```python
# Doing concatenation vertically and horizontally

# 1. Vertical Concatenation
print("\nVERTICAL CONCATENATION")
vertical = pd.concat([employee, employee2], axis=0, ignore_index=True)
print(vertical)


# 2. Horizontal Concatenation
print("\nHORIZONTAL CONCATENATION")
horizontal = pd.concat([employee, employee2], axis=1)
print(horizontal)
```

```python
print("\nFILTER - FINANCE, HR AND SALARY RANGE")

result = merged[
    (merged['Department'] == 'Finance') &
    (merged['Designation'] == 'HR') &
    (merged['Salary'].between(50000, 70000))
]

print(result)
```
